```python
import numpy as np

arr = np.arange(1, 10)

matrix = arr.reshape(3, 3)

cube = np.arange(1, 25).reshape(2, 3, 4)

print("Original:", arr)

print("\n3 x 3:\n", matrix)

print("\n3D shape:", cube.shape)

print(cube)
```
