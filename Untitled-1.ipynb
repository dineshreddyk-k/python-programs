```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Average:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
print("Count:", np.size(arr))
```

```python
matrix = np . array([[1,2,3],[4,5,6]])
print("cloumn sums (axis=0:)",np.sum(matrix,axis=0))
```

```python
arr=np.array([12,34,56,78,90])
print("arr > 20:",arr > 20)
print("values > 20:",arr[arr > 20])
```

```python
x=np.array([[1,2],[3,4]])
y=np.array([[5,6],[7,8]])
print("\nvertical stack:\n",np.vstack([x,y]))
print("\nhorizontal stack:\n",np.hstack([x,y]))
```

```python
rng=np.random.default_rng(42)
print("random integers:",rng.integers(1,20, size=5))
print("random float:",rng.random(5))
```

```python
A=np.array([[1,2],[3,5]])
B=np.array([[4,6],[8,9]])
print("element-wise multiplication:\n",A*B)
print("\nmatric multiplication:\n",A@B)
```
